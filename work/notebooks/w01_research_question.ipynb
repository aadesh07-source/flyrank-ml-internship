## 2) The question

### Search question

Can the available data help us identify records that may require greater
attention, so that a person or team can make better decisions about which
cases should be prioritized?

### Decision

The decision is whether a particular record should be given higher
priority for further review or action.

### Action

If the analysis identifies a case as high-priority, the person or team
could review the case first, investigate the underlying factors, and take
an appropriate action based on the available information.

### Cost of a wrong call

A false positive could result in time and resources being spent reviewing
a case that did not actually require immediate attention.

A false negative could result in an important case being overlooked and
not receiving attention when it was needed.

At this stage, I expect the false negative to be the more important error
because overlooking a genuinely important case could lead to a missed
opportunity for timely action.
